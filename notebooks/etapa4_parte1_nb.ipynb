# Etapa 4 — Parte 1: MLE, MAP e Naive Bayes

Esta parte individual implementa os estimadores Bernoulli de máxima verossimilhança (MLE), MAP com prior Beta(2,2) e Naive Bayes do zero. **O algoritmo EM pertence à Parte 2 (outro integrante).**

**Preparação:** no GitHub, o arquivo `trabalho_etapas_3_4_bayes.py` fica na raiz e este notebook em `notebooks/`. Coloque `auto_infracao_2021.csv` ao lado do `.py`. No Google Colab, envie o `.py` e o CSV para `/content`, mesmo que abra o notebook diretamente pelo GitHub.

```python
# Importa o módulo localizado na raiz do projeto ou em /content (Colab).
from pathlib import Path
import sys

NOME_MODULO = 'trabalho_etapas_3_4_bayes'
NOME_ARQUIVO = NOME_MODULO + '.py'
pastas_possiveis = [Path.cwd(), Path.cwd().parent, Path('/content')]
pasta_projeto = next((p for p in pastas_possiveis if (p / NOME_ARQUIVO).is_file()), None)

if pasta_projeto is None:
    raise FileNotFoundError(
        f'Arquivo {NOME_ARQUIVO} não localizado. No Colab, envie o .py e o CSV para /content; '
        'no repositório, mantenha o .py na raiz e os notebooks em notebooks/.'
    )
if not (pasta_projeto / 'auto_infracao_2021.csv').is_file():
    raise FileNotFoundError(
        f'CSV auto_infracao_2021.csv não encontrado em {pasta_projeto}. '
        'Coloque o CSV na mesma pasta do arquivo .py.'
    )
if str(pasta_projeto) not in sys.path:
    sys.path.insert(0, str(pasta_projeto))

import trabalho_etapas_3_4_bayes as projeto

import numpy as np
from sklearn.metrics import precision_score, recall_score, f1_score


projeto.demonstrar_mle_map(projeto.df_treino)
projeto.verificar_alinhamento_etapa1()
```

```python
modelo = projeto.NaiveBayesRaiz().fit(projeto.X_tr, projeto.y_tr)
acuracias, baselines = projeto.validacao_cruzada(projeto.df_treino_bruto, k=5)
previsoes = modelo.predict(projeto.X_te)
print("\nETAPA 4 — NAIVE BAYES DO ZERO")
print(f"Validação cruzada: {np.mean(acuracias):.2%} ± {np.std(acuracias, ddof=1):.2%}")
print(f"Baseline: {np.mean(baselines):.2%}")
print(f"Acurácia externa: {projeto.acuracia(projeto.y_te, previsoes):.2%}")
print(f"Precisão: {precision_score(projeto.y_te, previsoes, zero_division=0):.2%}")
print(f"Recall: {recall_score(projeto.y_te, previsoes, zero_division=0):.2%}")
print(f"F1: {f1_score(projeto.y_te, previsoes, zero_division=0):.2%}")
```
