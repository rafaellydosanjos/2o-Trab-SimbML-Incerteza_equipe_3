# Etapa 4 — Parte 1: MLE, MAP e Naive Bayes

Esta parte individual implementa os estimadores Bernoulli de máxima verossimilhança (MLE), MAP com prior Beta(2,2) e Naive Bayes do zero. **O algoritmo EM pertence à Parte 2 (outro integrante).**

**Preparação:** coloque este notebook, `trabalho_etapas_3_4_v6_1_bayes_corrigido.py` e `auto_infracao_2021.csv` na mesma pasta (no Colab: `/content`).

```python
import numpy as np
from sklearn.metrics import precision_score, recall_score, f1_score
import trabalho_etapas_3_4_v6_1_bayes_corrigido as projeto

projeto.demonstrar_mle_map(projeto.df_treino)
projeto.verificar_alinhamento_etapa1()
```

```python
modelo = projeto.NaiveBayesRaiz().fit(projeto.X_tr, projeto.y_tr)
acuracias, baselines = projeto.validacao_cruzada(projeto.df_treino_bruto, k=5)
previsoes = modelo.predict(projeto.X_te)
print("\nETAPA 4 — NAIVE BAYES DO ZERO")
print(f"Validação cruzada: {np.mean(acuracias):.2%} ± {np.std(acuracias, ddof=1):.2%}")
print(f"Baseline: {np.mean(baselines):.2%}")
print(f"Acurácia externa: {projeto.acuracia(projeto.y_te, previsoes):.2%}")
print(f"Precisão: {precision_score(projeto.y_te, previsoes, zero_division=0):.2%}")
print(f"Recall: {recall_score(projeto.y_te, previsoes, zero_division=0):.2%}")
print(f"F1: {f1_score(projeto.y_te, previsoes, zero_division=0):.2%}")
```
